```python
def minCostClimbingStairs(cost) -> int:
    cost.append(0)
    n = len(cost)
    #Start from the second-to-last element and iterate in reverse.
    for i in range(n-4,-1,-1):
        #Update each element with the minimum cost of either taking one step or two steps to reach the top.
        print(cost)
        cost[i] += min(cost[i+1], cost[i+2])
        print(cost)
    return min(cost[0], cost[1])
```

```python
matrix = [[1,2,3],[4,5,6],[7,8,9]]
row, col= len(matrix), len(matrix[0])
up, left = 0,0
down,right = row-1, col-1

out = []
while len(out) < row*col:
    for j in range(left, right+1):
        out.append(matrix[up][j])
    for i in range(up+1, down+1):
        out.append(matrix[i][right])
    if up!=down:
        for j in range(right-1, left-1, -1):
            out.append(matrix[down][j])
    if right!=left:
        for i in range(down-1, up, -1):
            out.append(matrix[i][left])
    left+=1
    right-=1
    up+=1
    down-=1

print(out)
```

```python
def fibonacci(n):
    num1, num2 = 0, 1
    if n == num1:
        return num1
    elif n == num2:
        return num2
    else:
        for _ in range(2, n+1):
            out = num1+num2
            num1 = num2
            num2 = out
        return out
```

```python
def rob(nums):
    n = len(nums)
    if n == 0:
        return 0
    
    dp = [0]*(n+1)
    dp[0] = 0
    dp[1] = nums[0]

    for i in range(2, n+1):
        dp[i] = max(dp[i-1], dp[i-2]+nums[i-1])
    return dp[n]
```

```python
def rob(nums):
    n = len(nums)
    if n == 0:
        return 0
    if n ==1:
        return nums[0]
    if n ==2:
        return max(nums[0], nums[1])
    

    def rob_line(arr):
        m = len(arr)
        if m == 0:
            return 0
        if m ==1:
            return arr[0]
        
        dp = [0]*(m+1)
        dp[0]=0
        dp[1]=arr[0]
        for i in range(2, m+1):
            dp[i] = max(dp[i-1], dp[i-2]+arr[i-1])
        return dp[m]
    caseA = rob_line(nums[:-1])
    caseB = rob_line(nums[1:])
    return max(caseA,caseB)
```

```python
def lpal(s):
    n = len(s)
    if n == 0:
        return ''
    if n == 1:
        return s
    
    dp=[[False]*(n) for _ in range(n)]

    st, longest = 0,1

    for j in range(n):
        for i in range(j-1,-1,-1):
            if s[i]==s[j]:
                if dp[i+1][j-1] or j-i+1==2:
                    dp[i][j] = True

                    currlen = j-i+1
                    if currlen > longest:
                        st = i
                        longest = currlen
    return s[st:st+longest]
```

```python
coins = [1,2,4,5]
amount = 15
max_val = amount + 1  # A value greater than any possible result
dp = [max_val] * (amount + 1)  # dp[x] = min coins to make amount x
dp[0] = 0  # Base case: 0 coins needed to make amount 0

for coin in coins:
    for x in range(coin, amount + 1):
        # Try using coin, and see if we get a better result
        dp[x] = min(dp[x], dp[x - coin] + 1)
    print(dp)
```

```python
nums = [1,2,3,4,5]

def subsets(nums):
    res,path = [],[]

    def dfs(i):
        if i >=len(nums):
            res.append(path[:].copy())
            return
        path.append(nums[i])
        dfs(i+1)
        path.pop()
        dfs(i+1)
    dfs(0)
    return res
```

```python
def setZeroes(self, matrix):
    if not matrix or not matrix[0]:
        return
    ROWS,COLS = len(matrix), len(matrix[0])
    fr = any(matrix[0][j] == 0 for j in range(COLS))
    fc = any(matrix[i][0] == 0 for i in range(ROWS))

    for i in range(1, ROWS):
        for j in range(1, COLS):
            if matrix[i][j] == 0:
                matrix[i][0] = 0
                matrix[0][j] = 0

    for i in range(1, ROWS):
        for j in range(1, COLS):
            if matrix[i][0] == 0 or matrix[0][j] == 0:
                matrix[i][j] = 0

    if fr:
        for j in range(COLS):
            matrix[0][j] = 0
    if fc:
        for i in range(ROWS):
            matrix[i][0] = 0
```
