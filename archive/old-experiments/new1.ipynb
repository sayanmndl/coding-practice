```python
from collections import deque

def topological_sort(num_nodes, edges):
    # 1. Calculate the in-degree for every node and build the adjacency list
    in_degree = {i: 0 for i in range(num_nodes)}
    adj_list = {i: [] for i in range(num_nodes)}
    
    for u, v in edges:
        # u must happen before v (u -> v)
        adj_list[u].append(v)
        in_degree[v] += 1
        
    # 2. Add all nodes with in-degree 0 to a queue
    queue = deque([node for node in in_degree if in_degree[node] == 0])
    
    result = []
    
    # 6. Repeat until the queue is empty
    while queue:
        # 3. Pull a node from the queue, add it to the result
        current = queue.popleft()
        result.append(current)
        
        # 4. For each neighbor of that node, decrement its in-degree
        for neighbor in adj_list[current]:
            in_degree[neighbor] -= 1
            
            # 5. If the in-degree hits 0, add it to the queue
            if in_degree[neighbor] == 0:
                queue.append(neighbor)
                
    # Cycle Check: If we processed all nodes, return the result. 
    # If not, it means there is a circular dependency.
    if len(result) == num_nodes:
        return result
    else:
        return "Cycle detected! No valid ordering exists."

# --- Example Usage ---

# 6 nodes (labeled 0 through 5)
nodes = 6

# Edges: [A, B] means A must be done before B
dependencies = [
    [5, 2],  # Node 5 points to 2
    [5, 0],
    [4, 0],
    [4, 1],
    [2, 3],
    [3, 1]
]

ordered_tasks = topological_sort(nodes, dependencies)
print(f"Valid task order: {ordered_tasks}")
```

```python
def solve_n_queens(n: int) -> list[list[int]]:
    solutions = []
    columns = set()
    diag1 = set()
    diag2 = set()

    def backtrack(row, placement):
        if row == n:
            solutions.append(placement[:])
            return

        for col in range(n):
            if col in columns or (row - col) in diag1 or (row+col) in diag2:
                continue

            columns.add(col)
            diag1.add(row-col)
            diag2.add(row+col)
            placement.append(col)

            backtrack(row+1, placement)

            columns.remove(col)
            diag1.remove(row-col)
            diag2.remove(row+col)
            placement.pop()
        
    backtrack(0,[])
    return solutions
```
